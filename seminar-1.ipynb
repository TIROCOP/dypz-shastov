```python
import sympy as sp

x = sp.symbols('x')
f = x**2 * sp.cos(x + 6) + x
df = sp.diff(f, x)
df
```

```python
y = sp.Function('y')
C1 = sp.symbols('C1')
general_solution = sp.Eq(y(x), 1 + C1/x)
general_solution
```

```python
sp.Eq(y(x), 1 + C1/x)
```

```python
initial_solution = sp.Eq(y(x), 1 + 1/x)
initial_solution
```

```python
sp.Eq(y(x), 1 + 1/x)
```

```python
exceptional_solution = sp.Eq(y(x), -1)
exceptional_solution
```

```python
import sympy as sp

x = sp.symbols('x')
f = x**2 * sp.cos(x + 6) + x
df = sp.diff(f, x)
df
```

```python
sp.diff(x**2 * sp.cos(x + 6) + x, x)
```

```python
y = sp.Function('y')
eq = sp.Eq(x * (y(x) + 1) * sp.diff(y(x), x), 1 - y(x)**2)
general_solution = sp.dsolve(eq)
general_solution
```

```python
sp.dsolve(
    sp.Eq(x * (y(x) + 1) * sp.diff(y(x), x), 1 - y(x)**2)
)
```

```python
initial_solution = sp.dsolve(
    eq,
    ics={y(1): 2}
)
initial_solution
```

```python
sp.dsolve(
    sp.Eq(x * (y(x) + 1) * sp.diff(y(x), x), 1 - y(x)**2),
    ics={y(1): 2}
)
```
